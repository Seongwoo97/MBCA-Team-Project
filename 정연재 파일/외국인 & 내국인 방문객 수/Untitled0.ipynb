```python
# 파일 읽어보기

import pandas as pd

외지인 = pd.read_csv('방문자수_히트맵_외지인_통합.csv')
현지인 = pd.read_csv('방문자수_히트맵_현지인_통합.csv')
외국인 = pd.read_csv('방문자수_히트맵_외국인_통합.csv')

print(외지인.shape)
print(외지인.columns)
display(외지인.head())
```

```python
# 컬럼명 통합(변경)

외지인 = 외지인.rename(columns={
    '기준연월': '연월',
    '기초지자체': '행정구명',
    '방문자 수': '외지인'
})

현지인 = 현지인.rename(columns={
    '기준연월': '연월',
    '기초지자체': '행정구명',
    '방문자 수': '현지인'
})

외국인 = 외국인.rename(columns={
    '기준연월': '연월',
    '지역': '행정구명',
    '외국인 방문자 수': '외국인'
})
```

```python
# 필요한 컬럼만 남기기
외지인 = 외지인[['연월', '행정구명', '외지인']]
현지인 = 현지인[['연월', '행정구명', '현지인']]
외국인 = 외국인[['연월', '행정구명', '외국인']]
```

```python
# 세 파일 합치기

방문자 = 외지인.merge(
    현지인,
    on=['연월', '행정구명'],
    how='inner'
)

방문자 = 방문자.merge(
    외국인,
    on=['연월', '행정구명'],
    how='inner'
)
```

```python
# 전체 방문자수 계산

방문자['전체'] = (
    방문자['외지인']
    + 방문자['현지인']
    + 방문자['외국인']
)
```

```python
# 컬럼 순서 변경

방문자 = 방문자[
    ['행정구명', '연월', '외지인', '현지인', '외국인', '전체']
]
```

```python
# 확인

print(방문자.columns)
display(방문자.head())
```

```python
print('행 수:', len(방문자))
print('행정구 수:', 방문자['행정구명'].nunique())
print('연월 수:', 방문자['연월'].nunique())
print('중복:', 방문자.duplicated(['행정구명', '연월']).sum())
```
