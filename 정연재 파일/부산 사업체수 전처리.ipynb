```python
import pandas as pd

df = pd.read_csv(
    '부산_사업자현황_통합_202201_202606_업종6분류.csv',
    encoding='utf-8-sig'
)

print(df.columns.tolist())
print(df.head())
```

```python
# 필요한 컬럼만 선택 + 컬럼명 변경
df = df[
    ['연월', '구분3', '업종대분류', '구분1', '당월①']
].copy()

df = df.rename(columns={
    '연월': '연도',
    '구분1': '행정구',
    '당월①': '사업체수'
})
```

```python
# 연월을 연도로 변경
df['연도'] = df['연도'].astype(str).str[:4].astype(int)
```

```python
# 사업체수 숫자형 변환
df['사업체수'] = (
    df['사업체수']
    .astype(str)
    .str.replace(',', '', regex=False)
    .astype(int)
)
```

```python
print(df.shape)
print(df.columns.tolist())
print(df.head(10))
```

```python
df.info()
```

```python
output_file = '부산_사업자현황_전처리.csv'

df.to_csv(
    output_file,
    index=False,
    encoding='utf-8-sig'
)

print(f'저장 완료: {output_file}')
```
