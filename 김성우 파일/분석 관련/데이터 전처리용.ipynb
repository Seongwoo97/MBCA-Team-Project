```python
df = pd.read_csv("/content/drive/MyDrive/MBCA/1차 팀 프로젝트/데이터 파일/busan_store_history.csv")
df.head()
```

```python
df['행정구'] = df['도로명주소'].str.extract(
    r'부산광역시\s+([가-힣]+(?:구|군))'
)

df[['도로명주소', '행정구']].head(10)
```

```python
df.head(10)
```

```python
df[['경도', '위도']] = (
    df['좌표']
    .str.extract(r'POINT\(([-\d.]+)\s+([-\d.]+)\)')
    .astype(float)
)

df[['좌표', '경도', '위도']].head(10)
```

```python
gu = gpd.read_file(
    '/content/drive/MyDrive/MBCA/1차 팀 프로젝트/데이터 파일/부산 시군구/BND_SIGUNGU_PG.shp'
)
```

```python
busan_gu = gu[
    gu['SIGUNGU_CD'].astype(str).str.startswith('21')
].copy()

busan_gu[['SIGUNGU_CD', 'SIGUNGU_NM']]
```

```python
busan_gu = busan_gu.to_crs(epsg=4326)
busan_gu.crs
```

```python
df[['경도','위도']].isna().sum()
```

```python
valid_df = df[
    df['경도'].notna() &
    df['위도'].notna()
].copy()
```

```python
import geopandas as gpd

gdf = gpd.GeoDataFrame(
    valid_df,
    geometry=gpd.points_from_xy(
        valid_df['경도'],
        valid_df['위도']
    ),
    crs='EPSG:4326'
)
```

```python
result = gpd.sjoin(
    gdf,
    busan_gu[['SIGUNGU_NM', 'geometry']],
    how='left',
    predicate='within'
)
```

```python
result = result.rename(
    columns={'SIGUNGU_NM': '행정구'}
)
```

```python
result[
    ['점포명', '도로명주소', '좌표', '행정구']
].head()
```

```python
result.to_csv('busan_store_history_v1.csv', index=False, encoding='utf-8-sig')
files.download('busan_store_history_v1.csv')
```
